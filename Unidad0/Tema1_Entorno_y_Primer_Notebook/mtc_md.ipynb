# Medidas de tendencia central

Son aquellas que nos permiten resumir un conjunto de datos en un solo valor representativo. Las medidas de tendencia central más comunes son:

1. **Media**: Es el promedio de todos los valores en un conjunto de datos. Se calcula sumando todos los valores y dividiendo entre el número total de valores.
   Fórmula:
   $$\displaystyle{\overline x = \frac{\sum_{i=1}^{n} x_i}{n}}$$
   Donde \( x_i \) son los valores del conjunto de datos y \( n \) es el número total de valores.

2. **Mediana**: Es el valor que se encuentra en el centro de un conjunto de datos ordenado. Si el número de valores es impar, la mediana es el valor del medio; si es par, es el promedio de los dos valores centrales.

3. **Moda**: Es el valor que aparece con mayor frecuencia en un conjunto de datos. Un conjunto de datos puede tener una moda, más de una moda (bimodal o multimodal) o ninguna moda si todos los valores son únicos.

pero eso es lo que se hace con un papel,por ejemplo, en un conjutno de 4 o 7 datos, sumas los datos, divides por el total, y si son... ¿17000 datos?, ¿lápiz y papel? No, para eso se utilizan herramientas computacionales que permiten manejar grandes cantidades de datos de manera eficiente. Programas como Excel, R, Python (con bibliotecas como Pandas y NumPy) son ideales para calcular medidas de tendencia central en conjuntos de datos extensos.

## Primer ejemplo práctico

La introducción es extensa, así que paciencia.

Comenzaremos utilizando un conjunto de datos pequeño,luego avanzaremos a conjuntos de datos más grandes. Supongamos que tenemos los siguientes datos de las edades de un grupo de personas, y finalmente a archivos en formato .csv con miles de registros.

```python
# Se debe importar la librería numpy, que es la librería con las funciones "NUM"éricas de "PY"thon
import numpy as np
datos = [
    23, 45, 12, 67, 34, 89, 54, 31, 22, 18,
    29, 41, 55, 63, 72, 81, 90, 15, 27, 38,
    49, 58, 66, 74, 82, 91, 10, 20, 30, 40,
    50, 60, 70, 80, 90
]
media = np.mean(datos)
print(f"La media es: {media}")
```

```python
# Se debe importar la librería statistics para usar la moda.
import statistics as st
moda = st.mode(datos)
print(f"La moda es: {moda}")
```

```python
mediana = np.median(datos)
print(f"La mediana es: {mediana}")
```

## Medidas de dispersión

Son las que nos permiten conocer la variabilidad o dispersión de un conjunto de datos. Las medidas de dispersión más comunes son:
1. **Rango**: Es la diferencia entre el valor máximo y el valor mínimo de un conjunto de datos.
   Fórmula:
   $$\displaystyle{Rango = x_{max} - x_{min}}$$
2. **Varianza**: Es una medida que indica cuánto se dispersan los valores de un conjunto de datos respecto a la media. Se calcula como el promedio de los cuadrados de las diferencias entre cada valor y la media.
   Fórmula:
   $$\displaystyle{\sigma^2 = \frac{\sum_{i=1}^{n} (x_i - \overline{x})^2}{n}}$$
   Donde \( \overline{x} \) es la media del conjunto de datos, \( x_i \) son los valores del conjunto de datos y \( n \) es el número total de valores.
3. **Desviación estándar**: Es la raíz cuadrada de la varianza y nos indica cuánto se desvían los valores de un conjunto de datos respecto a la media.
   Fórmula:
   $$\displaystyle{\sigma = \sqrt{\sigma^2}}$$


Ojo porque esas fórmulas, acá, no nos sirven para absolutamente nada jajaja, ni siquiera para entender como funciona esto en python, porque no vamos a hacer cálculos a mano, sino que vamos a utilizar funciones que ya están implementadas en librerías de python. Por ejemplo, para calcular la media, varianza y desviación estándar, podemos utilizar las funciones `mean()`, `var()` y `std()` de la librería **numpy**.

```python
rango = np.max(datos) - np.min(datos)
print(f"El rango es: {rango}")
```

```python
varianza = np.var(datos)
print(f"La varianza es: {varianza}")
```

```python
desv_estand = np.std(datos)
print(f"La desviación estándar es: {desv_estand}")
```
